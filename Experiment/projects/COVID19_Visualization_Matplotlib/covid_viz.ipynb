```
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    'Day': range(1,8),
    'Cases':[100,200,150,300,250,400,350],
    'Deaths':[1,3,2,4,2,5,3]
})

plt.plot(data['Day'], data['Cases'], label='Cases', marker='o')
plt.plot(data['Day'], data['Deaths'], label='Deaths', marker='x')
plt.title("COVID-19 Trend")
plt.xlabel("Day")
plt.ylabel("Count")
plt.legend()
plt.show()
```
