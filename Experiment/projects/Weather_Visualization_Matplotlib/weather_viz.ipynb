```
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    'Day': range(1,8),
    'Temperature': [30,32,33,31,29,28,30],
    'Rainfall': [5,0,0,10,0,0,2]
})

plt.plot(data['Day'], data['Temperature'], marker='o', label='Temp (C)')
plt.bar(data['Day'], data['Rainfall'], alpha=0.3, label='Rainfall (mm)')
plt.xlabel("Day")
plt.ylabel("Value")
plt.title("Weekly Weather")
plt.legend()
plt.show()
```
