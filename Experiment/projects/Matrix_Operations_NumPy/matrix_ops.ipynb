```
import numpy as np
import matplotlib.pyplot as plt

A = np.random.randint(1,10,(3,3))
B = np.random.randint(1,10,(3,3))

print("Matrix A:\n", A)
print("Matrix B:\n", B)

C = A + B
print("A + B:\n", C)

D = np.dot(A,B)
print("A x B:\n", D)

plt.imshow(D, cmap='viridis')
plt.colorbar()
plt.title("Matrix Product Heatmap")
plt.show()
```
