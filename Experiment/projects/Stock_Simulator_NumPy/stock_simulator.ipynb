```
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
S0 = 100
mu = 0.1
sigma = 0.2
T = 1
dt = 1/252
N = int(T/dt)

for i in range(5):
    prices = [S0]
    for _ in range(N):
        prices.append(prices[-1] * np.exp((mu - 0.5*sigma**2)*dt + sigma*np.sqrt(dt)*np.random.randn()))
    plt.plot(prices)
plt.title("Simulated Stock Prices")
plt.show()
```
